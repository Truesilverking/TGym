# TGym

Notebook principal para preparar, ejecutar, probar y generar TGym. La aplicación conserva su estructura React + Capacitor; este notebook centraliza las operaciones habituales.

```python
from pathlib import Path
import os, subprocess
PROJECT = Path.cwd()
FRONTEND = PROJECT / 'frontend'
assert (FRONTEND / 'package.json').exists(), 'Abre este notebook desde la carpeta principal de TGym.'
print('Proyecto:', PROJECT)
```

## Instalar las dependencias

```python
subprocess.run(['npm', 'ci'], cwd=FRONTEND, check=True, shell=os.name == 'nt')
```

## Verificar TGym

```python
subprocess.run(['npm', 'test'], cwd=FRONTEND, check=True, shell=os.name == 'nt')
subprocess.run(['npm', 'run', 'build'], cwd=FRONTEND, check=True, shell=os.name == 'nt')
print('Pruebas y compilación completadas.')
```

## Abrir TGym en modo de desarrollo

Esta celda permanece activa mientras la aplicación está abierta. Detén la celda para cerrar el servidor.

```python
subprocess.run(['npm', 'run', 'dev', '--', '--host', '127.0.0.1'], cwd=FRONTEND, check=True, shell=os.name == 'nt')
```

## Preparar el proyecto Android

```python
subprocess.run(['npm', 'run', 'build:mobile'], cwd=FRONTEND, check=True, shell=os.name == 'nt')
print('Proyecto Android actualizado en frontend/android.')
```
